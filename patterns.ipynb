```python
n=int(input())
for i in range(1,n+1):
    for j in range(1,n+1):
        print("*",end=" ")
    print()
```

```python
# * Piramid Pattern
n=int(input())
for i in range(1,n+1):
    for j in range(1,n-i+1):
        print(" ",end=" ")
    for k in range(1,i+1):
        print("*",end=" ")
    print()
```

```python
# * triangle pattern
n=int(input())
for i in range(1,n+1):
    for j in range(1,i+1):
        print("*",end=" ")
    print()
    
```

```python
# Full Piramid Triangle
n=int(input())
for i in range(1,n+1):
    for j in range(1,n-i+1):
        print(" ",end=" ")
    for k in range(1,i+1):
        print("*",end=" ")
    for l in range(1,i):
        print("*",end=" ")
    print()
    
```

```python
# Matrix triangle in odd numbers pattern 
n=int(input()) 
for i in range(1,n+1):
    for j in range(1,n+1):
        if j <= i:
            print(2*i-1,end=" ")
        else:
            print(" ",end=" ")
    print()
```

```python
# Matrix full pyramid triangle in odd numbers pattern 
n=int(input()) 
for i in range(1,n+1):
    for j in range(1,n-i+1):
        print(" ",end=" ")
    for k in range(1,i+1):
        print(2*i-1,end=" ")
    for l in range(1,i):
        print(2*i-1,end=" ")
    print()
    
```

```python
#Upside down triangle with number pattern
n=int(input())
for i in range(n,0,-1):
    for j in range(1,i+1):
        print(i,end=" ")
    print()
    
```

```python
# Number triangle on mirror view
n=int(input())
for i in range(1,n+1):
    for j in range(1,n-i+1):
        print(" ",end=" ")
    for k in range(1,i+1):
        print(k,end=" ")
    print()
```
